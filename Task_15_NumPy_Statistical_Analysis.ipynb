```python

import numpy as np

# Create a numerical dataset
marks = np.array([75, 80, 85, 90, 65, 70, 95, 88, 78, 82])

# Calculate statistical values
mean = np.mean(marks)
median = np.median(marks)
std_dev = np.std(marks)
minimum = np.min(marks)
maximum = np.max(marks)

# Display results
print("Student Marks:", marks)
print("Mean:", mean)
print("Median:", median)
print("Standard Deviation:", round(std_dev, 2))
print("Minimum Marks:", minimum)
print("Maximum Marks:", maximum)

# Compare mean and median
print("\nComparison:")
if mean > median:
    print("Mean is greater than Median")
elif mean < median:
    print("Mean is less than Median")
else:
    print("Mean and Median are equal")

# Identify unusually high and low values
print("\nUnusually High Values:")
print(marks[marks > mean + std_dev])

print("\nUnusually Low Values:")
print(marks[marks < mean - std_dev])
```
